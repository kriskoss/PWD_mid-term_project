##### Functions - plotting

```python
def convert_wide_to_long_forma_WORDL_BANK_dat(df, y_label="y_lable_missing",start_year=1960, end_year=2023):
    """ PLOTTING - converts form wide to long format the datagrame downloaded from the World Bank data website"""
    df_copy = df
    df_melted = df_copy.melt(
        id_vars=['Country Name', 'Country Code', 'Indicator Name', 'Indicator Code'], 
        value_vars=[str(i) for i in range(start_year, end_year)], 
        var_name='Year', 
        value_name=y_label  # This is now a unique name
    )

    # Converting 'Year' to integer to enalble the correct plot
    df_melted['Year'] = df_melted['Year'].astype(int)

    return df_melted
```

```python
def plot_worldbank_long_format(df, to_display_list, y_label ="y_lable_missing"):
    df_selected_countries = df[df['Country Name'].isin(to_display_list )]

    ### Plotting PSGR(air passenger numbers) vs year.
    g=sns.relplot(kind="line",x="Year",y=y_label, hue="Country Name",errorbar=None, data=df_selected_countries )
    ax = g.axes[0, 0]

    # Reformatting y axis to have have comma every 3 signinficant digits
    ax.yaxis.set_major_formatter(mticker.StrMethodFormatter('{x:,.0f}'))
    plt.show()
```

TBelow is the function responsible for downloading the file from the internet or loading backup file and feed the data into the Pandas dataframe. 

```python
def download_CSV_and_return_dataframe(URL, filename):
    try: # Tries to download the file - if successful the file will be saved in the "Data/" folder
        response = requests.get(URL) 
        folder= "Data/"
        
        with open(folder+filename, 'wb') as f:
            f.write(response.content)
        
        print_file_downloaded_msg(folder+filename)


    except Exception as e:
        # The connection with URL was not possible - source folder changed to the backup one
        folder = "Backup_data/"
        print_download_error_msg(folder+filename,e)
    
    
    try: # The source "folder" is used to upload the file into the dataframe
        return pd.read_csv(folder+filename)    
    
    except Exception as e:
        # The file read by Pandas is corrupted - it happens when the URL turns out to be incorrect
        # Reverts to backup folder for the data.
        print_red(f'{str(e)}\nUsing backup data: {folder+filename}!')
        folder = "Backup_data/"
                  
    return pd.read_csv(folder+filename)    
```

##### WORLD BANK - Passengers traffic per country (PSGR)
https://data.worldbank.org/indicator/IS.AIR.PSGR

In the case of downloading data from the World Bank, the process is similar to that of airports and runways data. However, there are a few key differences. Firstly, the downloaded file from the URL is in ZIP format, and it needs to be extracted. Additionally, the ZIP file contains three files, of which only one contains the actual data. The functions provided below handle the process of extracting the data from the ZIP file and transforming it into a dataframe.

```python
def return_data_filename__worldbank(search_string, data_folder):
    """The data downloaded from the World Bank website is in the form of a ZIP file. 
    Once extracted, it consists of two metadata files and one CSV file containing the data. 
    This function retrieves the name of the CSV file."""
    
    files = shutil.os.listdir(data_folder)
    for f_name in files:
        if re.search(search_string, f_name) and not re.match(r"Metadata", f_name):
            return f_name
    print_red("ERROR: No match.")
    return None
```

```python
def download_WorldBank_ZIP_and_return_dataframe(URL, params, filename, key_string):
    """
    This function attempts to download the dataset files in the form of a ZIP file from the World Bank website.
    If the download fails, it falls back to using files stored in the 'Backup_data' folder. In both cases, the ZIP file will be extracted to the 'Data/' folder, with a separate folder created for each ZIP file.

    The 'key_string' parameter (str) is used to locate the dataset's CSV file once the ZIP file is extracted. 
    Each ZIP file contains two metadata files and one file containing the data.

    The function" creates a file with the specified 'filename' and 
    returns the Pandas dataframe.
    """
    
    try: # Tries to download and save the file
        response = requests.get(URL,params) 
        folder= "Data/"
        with open(folder+filename, 'wb') as f:
            f.write(response.content)
        
        print_file_downloaded_msg(folder+filename)

    except Exception as e: # Download failed - reverting to the backup folder for data
        folder= "Backup_data/"
        print_download_error_msg(folder+filename,e)
    
    
    # Make Data/Extracted__ folder
    extracted_folder = "Data/Extracted__" +filename[:-4] 
    os.makedirs(extracted_folder, exist_ok=True) 

    # Unzip ZIP file into Data/Extracted__ folder
    if os.path.exists(folder + filename):
        try: ## Tries to unzip ZIP file 
            shutil.unpack_archive(folder + filename, extracted_folder,'zip')     
        
            print(f'"{filename}" extracted into {extracted_folder} ' + add_func_name())
        except Exception as e: # Extract failed - ZIP file is corrupted, maybe the URL is incorrect
            
            folder= "Backup_data/"
            print_red(f'{str(e)}\nReverting to backup ZIP file {folder + filename}!')
            shutil.unpack_archive(folder + filename, extracted_folder,'zip')     
    ###########


    # Finding the CSV data file and returning its file name
    CSV_filename = return_data_filename__worldbank(key_string, extracted_folder)

    # Constructing the file path by joining the extract folder and the data filename

    return pd.read_csv(f"{extracted_folder}/{CSV_filename}", skiprows=3)
    
    
```

```python
URL='https://api.worldbank.org/v2/en/indicator/IS.AIR.PSGR'
filename = "worldbank_PSGR.zip"
worldbank_PSGR_raw = download_WorldBank_ZIP_and_return_dataframe(URL,params,filename, "PSGR")
del URL, filename
worldbank_PSGR_raw.shape
```

##### Functions related to the data analysis

```python
def get_series_for_selected_year(df,year):
    """
    Extracts a column data in a form of a pd.Series from a World Bank data DataFrame for a given year. 
    The index of the Series is set as the 'Country Name' column of the DataFrame. 
    This function is designed to work specifically with certain World Bank CSV files.
    """
    year = str(year)
    
    series = pd.Series(df[year])
    series.index = df['Country Name']
    
    # Convert values to int and handle NaN values
    series = series.apply(lambda x: int(x) if pd.notna(x) else x)
    
    return series
```

```python
def print_world_population_factor(world_pop_in_year, pop_column, prnt=True):
    """
    Calculates the factor between the sum of the population for countries given a Series or DataFrame column and the total world population. 
    Optionally, it can print the output in a formatted way. """
    factor = pop_column.sum()/world_pop_in_year
    
    if prnt:
        print('{:.0%}'.format(factor)+f' of world population in year {year}', add_func_name())
    
    return factor   
```

#### The generate_align_factor function <a id='generate_align_factor_funtion'></a> 🔗
The **generate_align_factor** function is the most important function in my project. It calculates the relationship between the data in colA and colB using an arbitrary reference country. For example, it calculates the ratio between the GDP of Canada and its Passenger traffic, and stores this ratio for later use. The factor for the reference country is set to 1.

Next, the align_factor is calculated for selected parameters (e.g., GDP and PSGR) for each country in relation to the reference country (Canada). This allows for a comparison of the relationship between GDP and PSGR for each country.

The function is designed in such a way that when the result of GDP/PSGR multiplied by the reference_ratio is above 1, the factor is the result of the equation. However, if it is below 1, the division is inverted, meaning PSGR is divided by GDP instead of the other way around, and a negative sign is applied. This implies that the align_factor falls within the range of 1 to infinity and -1 to negative infinity.
    

```python
def generate_value(row, colA, colB, reference_ratio):
    """Internal functon for generate_align_factor function below"""
    if row[colA]/(row[colB]*reference_ratio) > 1:
        align_factor = 1+(row[colA] / (row[colB] * -reference_ratio))
    else:
        align_factor = -1+(row[colB] / row[colA] * reference_ratio)
    return align_factor.round(2)

def generate_align_factor(df,colA,colB,ref_country=None,outputColName=None,): #generate_align_factor
    """
    df - that need to contain col A and B and index that is 'Country Name
    colA,colB - column names to calculate the factor (e.g. GDP_2018, SURF_2018)
    ref - reference "Country Name" (e.g Canada)
    
    returns: df,reference_ratio, numA, numB
    numA,numB - those are values for colA and colB of the reference country
    """
    # Compare
    if outputColName:
        newName=outputColName
    else:
        partA = re.sub("_\d{4}", "", colA)
        newName = '_'.join([partA,'to',colB])
    
    # Calculate reference's country a reference ratio (GDP/passengers)
    # numA and numB are the values for reference country (i.e. Canada) and the respective columns(i.e. GDP and PSGR)
    if ref_country:
        
        numA = df.loc[ref_country, colA]
        numB = df.loc[ref_country, colB]
        reference_ratio = numA / numB
        
    else: 
        numA = None
        numB = None
        reference_ratio=1
    
    # Calculate ratio for each country and divide by reference ratio
    df[newName] = df.apply(generate_value, args=(colA, colB, reference_ratio), axis=1)
    
    return df,reference_ratio, numA, numB

```

```python
## Stores dropped rows for which factor is exceeded
def exclude_to_high_factor_countries(df,factor_col_name, factor):
    """ 
   This function filters out countries that fall outside the align factor 
   (check generate_align_factor function - above)
    
   Returns the:
            - the original dataframe, 
            - the adjusted dataframe - the data frame which do not contain any more countries below the threshold
            - the dataframe of dropped rows
    """
    countries_to_exclude = df[(df[factor_col_name]< -factor) | (df[factor_col_name]> factor)
        ].index.tolist()
    
    df_copy = df.copy() # To avoid modification of the original data
    
    # Drop the rows
    df_adjusted = df_copy.drop(countries_to_exclude)
    
    # Creating a dataframe of dropped rows
    dropped_countries_df = df_copy.loc[countries_to_exclude]
    
   
    return df, df_adjusted, dropped_countries_df
```

```python
def df_row__float_to_int(df,index_name):
    """Converts type of the data in a row from float to int"""
    # Dataframe has to be transposed to change the type of the row.
    df_t = df.transpose()
    df_t[index_name] = df_t[index_name].astype(int)
    df = df_t.transpose()
    
    return df
```

```python
def df_row__float_to_percent(df, index_name, decimal_pts=0): #df_row__float_to_percent
    """
    Converts the values in a specific row of dataframe to %.
    
    df(Dataframe) -the input dataframe.
    index_name (str): The name of the row to convert to percentages.
    decimal_pts (int, optional): The default is 0 . Number of decimal points to round the percentages to.,

    Returns converted dataframe
    """
    df_t = df.transpose()

    # Define a conversion function that handles different cases based on value
    def convert_to_percentage(x):
        if abs(x) > 0.01 :
            return str(int(x * 100)) + '%'
        else:
            return str(round(x * 100, 1)) + '%'
        
    # Convert the specified row values to percentages
    df_t[index_name] = df_t[index_name].apply(convert_to_percentage)

    # Transpose back to the original structure
    df = df_t.transpose()
    
    return df
```

#### plot_one_vs_other function
The function **plot_one_vs_other** is used to create a plot comparing two values, such as GDP to PSGR as shown in earlier examples. The key aspect of this function is to display the data in a specific manner where the **reference country** has two bars of equal height: one representing GDP and the other representing PSGR. This visual representation corresponds to an align_factor of 1. Additionally, the maximum y-limits for both the left and right scales need to be adjusted to ensure that the highest values of colA (e.g., GDP) and colB (e.g., PSGR) are within the plot area.

When modifying the y-limit of the left scale, the y-limit of the right scale should be adjusted by the same factor. This ensures consistent scaling between the two values.

The reference country bars are represented by semi-transparent blue and red colors. When these colors are combined, they result in a magenta color for the bars. If any other colors are visible, it indicates an error in the scale factors used, specifically the reference ratio.

```python
def plot_one_vs_other(df, colA, colB, ref_country, ref_LEFT, ref_RIGHT,colorA="darkslateblue", colorB="orange", title=None):
    """
    df - dataframe that need to contain index "Country Name"
    colA,colB - column names which will be plotted next to each other (e.g GDP_2018, PSGR_2018
    ref_country - ref_countryerence country name (e.g "Canada")
    """
    # Create a list of colors for GDP plot
    colA_colors = ['red' if index == ref_country else colorA for index in df.index]
    colB_colors = ['blue' if index == ref_country else colorB for index in df.index]

    # Create a bar plot for GDP
    fig, ax1 = plt.subplots(figsize=(14, 8))
    df[colA].plot(kind='bar', color=colA_colors, ax=ax1, label=colA)

    # Create a new y-axis for PSGR on the right side
    ax2 = ax1.twinx()
    df[colB].plot(kind='bar', color=colB_colors, ax=ax2, alpha=0.5, label=colB)

    # Set title and labels
    if not title:
        title = f'{colA} vs {colB} in {year}'
    plt.title(title)
    ax1.set_ylabel(f'{colA} in [units]')
    ax2.set_ylabel(f'{colB} in [units]')

 
    # Setting the upper limit of left and right plots
    colA_factor = df[colA].max()/ref_LEFT
    colB_factor = df[colB].max()/ref_RIGHT
    
  
    # Factor used depends if factor for colA or colB is larger - the larger factor will be then used on both, left and right scales
    if colA_factor < colB_factor:
        factor = colB_factor
    else:
        factor= colA_factor

    ax1.set_ylim(0, ref_LEFT*factor*1.1)
    ax2.set_ylim(0, ref_RIGHT*factor*1.1)
        

    # Change color of ref_country label to red
    for label in ax1.get_xticklabels():
        if label.get_text() == ref_country:
            label.set_color('red')

    # Add legends
    ax1.legend(loc='upper left')
    ax2.legend(loc='upper right')
```

```python
def str_to_percentage(key):
    return str(int(round(float(key) * 100)))+'%'
```

```python
def keep_selected_columns(df, *cols_to_keep): #keep_selected_columns
    # This function removes columns that are not specified in the arguments list
    new_df = df[[col for col in df.columns if col in cols_to_keep]]
    new_df = new_df.reindex(columns=cols_to_keep)
    return new_df
```

```python
def get_stats(df,col_name, world_pop, pop_col_name):
    """Extracts some statistical values from the selected column and returns data combined in the Series"""
    series = df[col_name]
    s_min = pd.Series([series.min()], index=["Min."])
    s_max = pd.Series([series.max()], index=['Max.'])
    s_std = pd.Series([series.std()], index=['Std.'])
    s_count = pd.Series([series.count()], index=["Count"])
    s_mean = pd.Series([series.mean()],index=['Mean'])

    ratio = print_world_population_factor(world_pop, df[pop_col_name],prnt =False)
    s_w_pop_ratio = pd.Series([ratio], index=["world_pop_ratio"])
    
    s_stats = pd.concat([s_min, s_max, s_std, s_count, s_mean, s_w_pop_ratio])
    
    return s_stats
```

```python
def df_row__float_to_int(df, index_name): 
    # Transpose the DataFrame
    df_t = df.transpose()

    # Convert the 'world_pop_fract' column to percentage, round to 0 decimal digits, and convert to string
    df_t[index_name] = df_t[index_name].apply(lambda x: int(x) if isinstance(x, float) else x)

    # Transpose back to the original structure
    df = df_t.transpose()
    
    return df
```

```python
def apply_to_all_rows_except(df, excluded_row):
    """
    Converts values in every row to the percentage except excluded row.
    
    df -   input DataFrame.
    excluded_row (str): The name of the row to exclude.

    returns converted dataframe
    
    """
    for row in df.index:
        if row != excluded_row:
            df = df_row__float_to_percent(df, row)

    return df
```

Through data analysis, it is evident that GDP serves as a highly reliable predictor of passenger traffic. This relationship is supported by various online articles, such as the one found at https://www.ishkaglobal.com/News/Article/5515/GDP-growth-vs-air-travel-A-conscious-uncoupling. 

Furthermore, it is interesting to note that factors like population and country size have limited impact on passenger traffic, despite the viability of air travel especially in large countries.

### Runways - example for further exploration

Below, there are two plots:

1. The first plot shows the distribution of runway directions worldwide.
2. The second plot focuses specifically on the United States, displaying the distribution of runway directions within the country.

```python
df = apt_and_rwys3
df = df[df['continent'] == 'EU']
df = df[df['le_heading_degT'].between(0, 180)]

# now create a histogram with seaborn
plt.figure(figsize=(10,6))
sns.histplot(df['le_heading_degT'], bins=180, kde=False)

plt.title('Distribution of le_heading_degT - WORLD')
plt.xlabel('le_heading_degT (in degrees)')
plt.ylabel('Count')
plt.show()
```

From the plot, it can be observed that the bars, representing runway direction, tend to accumulate near the middle and decrease in length as they approach the 0 and 180-degree ends. I chose to use the range of 0 to 180 degrees only, excluding 360 degrees, as including the full circle of data would be redundant. Each runway has two ends, so this plot displays only the "low value" end.

Regarding the accumulation near the center, one possible reason is the prevailing wind direction on the planet. The Earth's rotation influences the global wind patterns, leading to the prevailing wind direction for atmospheric air masses in the mid-latitudes being generally from the west. This is influenced by a the Coriolis effect(the apparent deflection of the path of an object that moves within a rotating coordinate system).

```python
df = apt_and_rwys4
df = df[df['country_name'] == 'United States']
# df = df[df['le_heading_degT'].between(0, 180)]

## FILTERING AIRPORTS WHICH HAS le heading and no he heading
df = df[df['le_heading_degT'].between(0, 180) & df['he_heading_degT'].notna()]
# df = df[df['he_heading_degT'].notna()]
print(df.shape)


# now create a histogram with seaborn
plt.figure(figsize=(10,6))
sns.histplot(df['le_heading_degT'], bins=180, kde=False)

plt.title('Distribution of le_heading_deg - US')
plt.xlabel('le_heading_degT (in degrees)')
plt.ylabel('Count')
plt.show()
```

Finally, the plot above shows the runway directions in the United States. It is quite common in the US to use cardinal directions for building streets or establishing state lines. This can be seen as spikes in the 0, 45, 90, 135, and 180-degree directions.
